```python
# SARIMA Model

# Import libraries

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from statsmodels.tsa.statespace.sarimax import SARIMAX

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
```

```python
# Store SARIMA results

s_results = []
```

```python
# Get all states

states = df['State'].unique()
```

```python
# Run SARIMA model for every state

for state in states:

    # Filter one state
    state_df = df[df['State'] == state]

    # Set Date as index
    state_df = state_df.set_index('Date')

    # Convert to weekly frequency
    state_df = state_df.asfreq('W')

    # Sales time series
    sales_series = state_df['Sales']

    # Train data
    train = sales_series[:-8]

    # Validation data
    valid = sales_series[-8:]

    # Build SARIMA model
    model = SARIMAX(
        train,
        order=(1,1,1),
        seasonal_order=(1,1,1,52)
    )

    # Fit model
    model_fit = model.fit(disp=False)

    # Forecast next 8 weeks
    forecast = model_fit.forecast(steps=8)

    # Evaluation metrics
    mae = mean_absolute_error(valid, forecast)

    rmse = np.sqrt(
        mean_squared_error(valid, forecast)
    )

    # Store results
    s_results.append([
        state,
        mae,
        rmse
    ])
```

```python
# Convert SARIMA results into table

s_results_df = pd.DataFrame(
    s_results,
    columns=['State', 'MAE', 'RMSE']
)

s_results_df
```

```python
# Average MAE and RMSE

print("Average MAE:", s_results_df['MAE'].mean())

print("Average RMSE:", s_results_df['RMSE'].mean())
```

```python
# Import libraries

import numpy as np
import pandas as pd

from prophet import Prophet

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
```

```python
# Store Prophet results

p_results = []
```

```python
import logging

logging.getLogger('cmdstanpy').disabled = True

import warnings
warnings.filterwarnings("ignore")
```

```python
# Run Prophet model for every state

for state in states:

    # Filter one state
    state_df = df[df['State'] == state].copy()

    # Select required columns
    state_df = state_df[['Date', 'Sales']]

    # Rename columns for Prophet
    state_df = state_df.rename(
        columns={
            'Date': 'ds',
            'Sales': 'y'
        }
    )

    # Train data
    train = state_df[:-8]

    # Validation data
    valid = state_df[-8:]

    # Create Prophet model
    model = Prophet()

    # Train model
    model.fit(train)

    # Create future dataframe
    future = model.make_future_dataframe(
        periods=8,
        freq='W'
    )

    # Forecast
    forecast = model.predict(future)

    # Last 8 predictions
    predictions = forecast['yhat'][-8:].values

    # Evaluation metrics
    mae = mean_absolute_error(
        valid['y'],
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            valid['y'],
            predictions
        )
    )

    # Store results
    p_results.append([
        state,
        mae,
        rmse
    ])
```

```python
 # Convert Prophet results into table

p_results_df = pd.DataFrame(
    p_results,
    columns=['State', 'MAE', 'RMSE']
)

p_results_df
```

```python
# Import libraries

import numpy as np
import pandas as pd

from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
```

```python
# Store XGBoost results

xgb_results = []
```

```python
# Run XGBoost for every state

for state in df['State'].unique():

    # Filter one state
    state_df = df[df['State'] == state]

    # Features
    X = state_df.drop(
        columns=['State', 'Date', 'Sales']
    )

    # Target
    y = state_df['Sales']

    # Train data
    X_train = X[:-8]
    y_train = y[:-8]

    # Validation data
    X_valid = X[-8:]
    y_valid = y[-8:]

    # Build model
    model = XGBRegressor()

    # Train model
    model.fit(X_train, y_train)

    # Forecast
    forecast = model.predict(X_valid)

    # Evaluation metrics
    mae = mean_absolute_error(
        y_valid,
        forecast
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_valid,
            forecast
        )
    )

    # Store results
    xgb_results.append([
        state,
        mae,
        rmse
    ])
```

```python
# Convert XGBoost results into table

xgb_results_df = pd.DataFrame(
    xgb_results,
    columns=['State', 'MAE', 'RMSE']
)

xgb_results_df
```

```python
# Import libraries

import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    LSTM,
    Dense
)
```

```python
# Store LSTM results

lstm_results = []
```

```python
# Run LSTM model for every state

for state in states:

    # Filter one state
    state_df = df[df['State'] == state].copy()

    # Sales values
    sales = state_df['Sales'].values.reshape(-1, 1)

    # Split train and validation
    train_sales = sales[:-8]

    valid_sales = sales[-8:]

    # Scale only training data
    scaler = MinMaxScaler()

    train_scaled = scaler.fit_transform(train_sales)

    valid_scaled = scaler.transform(valid_sales)

    # Train sequences

    X_train = []

    y_train = []

    for i in range(30, len(train_scaled)):

        X_train.append(
            train_scaled[i-30:i]
        )

        y_train.append(
            train_scaled[i]
        )

    X_train = np.array(X_train)

    y_train = np.array(y_train)

    # Validation sequences

    X_valid = []

    y_valid = []

    combined = np.concatenate(
        [train_scaled[-30:], valid_scaled]
    )

    for i in range(30, len(combined)):

        X_valid.append(
            combined[i-30:i]
        )

        y_valid.append(
            combined[i]
        )

    X_valid = np.array(X_valid)

    y_valid = np.array(y_valid)

    # Build model

    model = Sequential()

    model.add(
        LSTM(
            50,
            activation='relu',
            input_shape=(X_train.shape[1], 1)
        )
    )

    model.add(Dense(1))

    # Compile model
    model.compile(
        optimizer='adam',
        loss='mse'
    )

    # Train model
    model.fit(
        X_train,
        y_train,
        epochs=20,
        batch_size=16,
        verbose=0
    )

    # Predictions
    predictions = model.predict(X_valid)

    # Inverse transform predictions
    predictions = scaler.inverse_transform(
        predictions
    )

    # Inverse transform actual values
    y_valid_actual = scaler.inverse_transform(
        y_valid
    )

    # Evaluation metrics
    mae = mean_absolute_error(
        y_valid_actual,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_valid_actual,
            predictions
        )
    )

    # Store results
    lstm_results.append([
        state,
        mae,
        rmse
    ])
```

```python
# Convert LSTM results into table

lstm_results_df = pd.DataFrame(
    lstm_results,
    columns=['State', 'MAE', 'RMSE']
)

lstm_results_df
```

```python
# Average MAE and RMSE

print("Average MAE:", lstm_results_df['MAE'].mean())

print("Average RMSE:", lstm_results_df['RMSE'].mean())
```

```python
# Store future forecasts

future_forecasts = []

# Get all states

states = df['State'].unique()
```

```python
# Run forecasting for every state

for state in states:

    # Filter state data
    state_df = df[df['State'] == state].copy()

    # Sales values
    sales = state_df['Sales'].values.reshape(-1, 1)

    # Scale data
    scaler = MinMaxScaler()

    scaled_sales = scaler.fit_transform(sales)

    # Create sequences
    X = []
    y = []

    for i in range(30, len(scaled_sales)):

        X.append(scaled_sales[i-30:i])

        y.append(scaled_sales[i])

    # Convert to numpy arrays
    X = np.array(X)

    y = np.array(y)

    # Build model
    model = Sequential()

    model.add(
        LSTM(
            50,
            activation='relu',
            input_shape=(30, 1)
        )
    )

    model.add(Dense(1))

    # Compile model
    model.compile(
        optimizer='adam',
        loss='mse'
    )

    # Train model
    model.fit(
        X,
        y,
        epochs=20,
        batch_size=16,
        verbose=0
    )

    # Last 30 values
    last_30 = scaled_sales[-30:]

    # Reshape input
    X_input = last_30.reshape(1, 30, 1)

    # Predict next 8 weeks
    future_predictions = []

    for i in range(8):

        pred = model.predict(
            X_input,
            verbose=0
        )

        future_predictions.append(
            pred[0][0]
        )

        # Update input
        X_input = np.append(
            X_input[:, 1:, :],
            [[[pred[0][0]]]],
            axis=1
        )

    # Inverse transform
    future_predictions = scaler.inverse_transform(
        np.array(future_predictions).reshape(-1, 1)
    )

    # Store results
    for week in range(8):

        future_forecasts.append([
            state,
            f"Week {week+1}",
            future_predictions[week][0]
        ])
```

```python
# Convert forecasts into dataframe

forecast_df = pd.DataFrame(
    future_forecasts,
    columns=[
        'State',
        'Week',
        'Forecasted_Sales'
    ]
)

# Display dataframe

forecast_df
```
